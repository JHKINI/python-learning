 📌 짝수 추출 + 개수 + 합 + 평균

```python
def study_1():
  nums = [1,2,3,4,5,6]
  r = []
  count = 0
  add = 0

  for n in nums:
    if n % 2 ==0:
      count +=1
      add += n
      r.append(n)
  average = add /count

  print(r)
  print(count)
  print(add)
  print(average)
study_1()
```

데이터 입력받기, 평균구하기오류처리

```python
def study_2():
  nums = list(map(int, input().split()))
  r = []
  count = 0
  add = 0

  for n in nums:
      if n % 3 ==0:
        count +=1
        add += n
        r.append(n)

  if count > 0:
      average = add /count
  else:
      average = 0

  print(r)
  print(count)
  print(add)
  print(average)
study_2()
```

클래스 작성

```python
def study_3():
  class Calc:
      def __init__(self, nums):
          self.nums = nums

      def even_info(self):
          count = 0
          r = []

          for n in self.nums:
              if n % 2 == 0:
                  count += 1
                  r.append(n)

          return r, count


  nums = [1, 2, 3, 4]
  obj = Calc(nums)

  result, count = obj.even_info()

  print(result)
  print(count)
study_3()
```

*args:몇 개 들어올지 모르는 값을 다 받기, isinstance:어떤 타입인지 확인

```python
def study_3():
    class 계산기:
        def __init__(self, *nums):
            self.숫자 = []

            for n in nums:
                if isinstance(n, int):
                    self.숫자.append(n)

        def 더하기(self):
            return sum(self.숫자)
study_3()
```
